```python
# Write a python program that, given an input list, will filter the input above a user defined threshold. 
# This is to be done with a standard function.
# That is, given a list [1,2,3,4,5,6,7,8,9], and an argument (6), it should return [1,2,3,4,5,6]

def filter_threshold(lst, threshold):
    result = []
    for z in lst:
        if isinstance(z, list):  # if the listed data here is a list within a list (sublist) then use recursion. 
            result.append(filter_threshold(z, threshold))
        else:                    # this is an integer 
            if z <= threshold:
                result.append(z) 
    return result 
    
# Here are a few examples

print(filter_threshold([1, [3, 6, 9], 18], 7))
```

```python
# look how 9 and 18 were dropped off the list -- now look at another one but further nesting 

print(filter_threshold([[[[1, 3, 10]]], 300, [9, [4, [8]]]], 67))
```
