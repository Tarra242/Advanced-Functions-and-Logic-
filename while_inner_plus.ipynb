```python
# Write a python program that, given an input list of any level of complexity/nestedness, will return the inner most list plus 1. 
# This is to be done with a while loop. Note: the input will contain only integers or lists.

def innermost_plus_one(nested_list):
    current = nested_list 
    while any(isinstance(z, list) for z in current):   # I keep forgetting the colons at the end of the lines---kept getting errors. Ive got it now! 
        for z in current:
            if isinstance(z, list):
                current = z 
                break    # the following list will only show a list of integers 
    return [z + 1 for z in current]

# Here are 2 different types of nesting loops in the easiest form to read. 

print(innermost_plus_one([2, 4, [6, 8, [10, 12]]]))   

print(innermost_plus_one([100, 200, 300]))            
```
